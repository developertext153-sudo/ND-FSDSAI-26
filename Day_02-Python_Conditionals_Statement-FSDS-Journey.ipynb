```python
# Advanced Electricity Bill Calculations
```

```python
cust_name=input("Enter your name : ")
unit_used=int(input("Enter the units : "))
per_unit_cost=3
fixed_charges=100

# calculating the electricity bill
if unit_used<0:
  print("Invalid Input")
else:
  if unit_used<=100:
    total_electricity_bill=unit_used*per_unit_cost+fixed_charges

  elif unit_used>=101 and unit_used<=200:
    per_unit_cost=5
    total_electricity_bill=100*3+(unit_used-100)*per_unit_cost+fixed_charges

  elif unit_used>=201 and unit_used<=300:
    per_unit_cost=7
    total_electricity_bill=100*3+100*5+(unit_used-200)*per_unit_cost+fixed_charges

  else:
    per_unit_cost=8
    total_electricity_bill=100*3+100*5+100*7+(unit_used-300)*per_unit_cost+fixed_charges

  print("\n--VVNL--")
  print(f'Customer Name: {cust_name}')
  print(f'Unit Used: {unit_used}')
  print(f'Per Unit Cost: {per_unit_cost}')
  print(f'Total Amount : {total_electricity_bill:.2f}')
  print("-------------------------------------------------")
```
