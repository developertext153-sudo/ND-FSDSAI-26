```python
department_expenses = {
    "HR": 25000,
    "IT": 75000,
    "Marketing": 40000,
    "Finance": 35000,
    "Operations": 25000
}
```

```python
# total expenses of all departments
total_expenses=0
for i in department_expenses:
    total_expenses+=department_expenses[i]
print(f'Total Expenses of all departments :{total_expenses}')
```

```python
# find the no of departments
department_count=0
for i in department_expenses:
    department_count+=1
print(f'Total Department :{department_count}')

# average expenses per day
average_expenses=round((total_expenses/department_count),2)
print(f'Average Expenses per day :{average_expenses}')

# each department percentage contribution in total_expenses and finding the total expenses
new_percentage_dict={}
for i in department_expenses:
    new_percentage_dict[i]=round((department_expenses[i]/total_expenses)*100,2)
    print(f'Percentage contribution {i} is : {(department_expenses[i]/total_expenses)*100}')
print(f'Total Expenses of all departments :{total_expenses}')
print(f'New Percentage Dictionary :{new_percentage_dict}')
```

```python

employees = (
    (101, "Rahul", "IT", 45000),
    (102, "Priya", "HR", 35000),
    (103, "Amit", "IT", 55000),
    (104, "Sneha", "Finance", 50000),
    (105, "Vikram", "IT", 60000),
    (106, "Neha", "HR", 40000),
    (107, "Rohit", "Finance", 48000)
)

# Exercise 1: Total and average monthly salary

total_salary = 0

for employee in employees:
    total_salary += employee[3]

total_employees = len(employees)
average_salary = total_salary / total_employees

print("----- Salary Summary -----")
print(f"Total Employees: {total_employees}")
print(f"Total Monthly Salary: {total_salary}")
print(f"Average Monthly Salary: {average_salary:.2f}")


# Exercise 2: Highest-paid employee

highest_paid = employees[0]

for employee in employees:
    if employee[3] > highest_paid[3]:
        highest_paid = employee

print("\n----- Highest Paid Employee -----")
print(f"Employee ID: {highest_paid[0]}")
print(f"Employee Name: {highest_paid[1]}")
print(f"Department: {highest_paid[2]}")
print(f"Salary: {highest_paid[3]}")


# Exercise 3: Department-wise employee counts

department_count = {}

for employee in employees:
    department = employee[2]

    if department in department_count:
        department_count[department] += 1
    else:
        department_count[department] = 1

print("\n----- Department-wise Employee Count -----")

for department, count in department_count.items():
    print(f"{department}: {count}")


# Exercise 4: Department-wise total salaries

department_salary = {}

for employee in employees:
    department = employee[2]
    salary = employee[3]

    if department in department_salary:
        department_salary[department] += salary
    else:
        department_salary[department] = salary

print("\n----- Department-wise Total Salary -----")

for department, salary in department_salary.items():
    print(f"{department}: {salary}")
```
