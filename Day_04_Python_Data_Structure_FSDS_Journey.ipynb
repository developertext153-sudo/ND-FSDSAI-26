```python
sales=[1200, 1500, 800, 2200, 1750, 900, 3000]
```

```python
sales
```

```python
# Excercise 1: find Sales
```

```python
# total sales
```

```python
total_sales=0
for i in sales:
    total_sales+=i
print("Total Sales:", total_sales)
```

```python
# average sales
sales_count=len(sales)
average_sales=total_sales/sales_count
print("Average Sales:", average_sales)
```

```python
# maximum sales
max_sales=max(sales)
```

```python
# minimum sales
min_sales=min(sales)
```

```python
# no of` sales greater than 1500
sales_greater_than_1500=0
for i in sales:
    if i > 1500:
        sales_greater_than_1500+=1
print("No of Sales greater than 1500:", sales_greater_than_1500)
```
