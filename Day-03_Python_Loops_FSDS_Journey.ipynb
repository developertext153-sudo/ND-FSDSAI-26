```python
# While Loop Continues as long as condition is True
```

```python
count=1 # initialization
while count<=5: # condition
  print(count) # execution
  count+=1 # updation
```

```python
for i in range(1,6): # means 1-5
  if i==3:
    break
  print(i)
```

```python
# continue in loop : skips the current iteration
```

```python
for i in range(1,6):
  if i%2==0:
    continue
  print(i)
```

```python
# Excercise 4: Factorial of a number
number=int(input("Enter the number : "))
fact=1
if number<0:
  print("Factorial does not exist for negative number")
elif number==0:
  print("Factorial of 0 is 1")
else:
  for i in range(1,number+1):
    fact=fact*i
  print(fact)
```

```python
# Mini Project - Daily Sales Analysis
sales=[1200,5000,2500,800,0,3000]
# calculating the total sales
total_sales=0

for i in sales:
  total_sales=float(total_sales+i)
print(f"Total sales is {total_sales}")

# counting the valid sales(greater than zero)
count_sales_=0
for i in sales:
  if i>0:
    count_sales_=count_sales_+1
print(f'Tota Sales count is {count_sales_}')

# calculating the average sales
average_sales=float(total_sales/len(sales))
print(f'Average sales is {round(average_sales,2)}')

# finding the largest sales
largest_sales=0.0
for i in sales:
  if i>largest_sales:
    largest_sales=i
print(f'Largest sales is {largest_sales}')

# counting the zero-sales records
zero_sale=0
for i in sales:
  if i==0:
    zero_sale=zero_sale+1
print(f'Count of zero sales records {zero_sale}')

# counting the sales greater than 3000
sale_greater3000=0
for i in sales:
  if i>3000:
    sale_greater3000=sale_greater3000+1
print(f'Count of 3000 sales records {sale_greater3000}')
```
