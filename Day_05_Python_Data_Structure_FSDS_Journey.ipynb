```python
# finding the total sales
sales=[1200, 1500, 800, 2200, 1750, 900, 3000]
total_sales=0
for i in range(len(sales)):
    total_sales+=sales[i]
print(f'Total Sales: {total_sales}')

# calculating average sales
average_sales=total_sales/len(sales)
print(f'Average Sales: {average_sales}')

# find the highest sales
highest_sales=sales[0]
for i in sales:
    if i>highest_sales:
        highest_sales=i
print(f'Highest Sales: {highest_sales}')

# find the lowest sales
lowest_sales=sales[0]
for i in sales:
    if i<lowest_sales:
        lowest_sales=i
print(f'Lowest Sales: {lowest_sales}')

# count the days when sales exceeded 3000
count=0
for i in sales:
    if i>3000:
        count+=1
print(f'Days when sales exceeded 3000: {count}')

# create a new list with sales greater than 3000
high_sales=[]
for i in sales:
    if i>3000:
        high_sales.append(i)
print(f'Sales greater than 3000: {high_sales}')

# Sort the sales in descending order
sorted_sales=sorted(sales,reverse=True)
print(f'Sorted Sales: {sorted_sales}')
        
```

```python
# Excercise 1.1 Second highest distinct sales
sales = [1200, 1500, 800, 2200, 1750, 900, 3000]

# Highest sales 
highest_sales=float('-inf')
for i in sales:
    if i>highest_sales:
        highest_sales=i
print(f'Highest Sales: {highest_sales}')

# Second highest sales 
second_highest_sales=float('-inf')
for i in sales:
    if i>second_highest_sales and i<highest_sales:
        second_highest_sales=i
print(f'Second Highest Sales: {second_highest_sales}')

# For disctinct sales, we can use set data structure to remove duplicates and then find the second highest sales.
# empty set creation
s1=set()

for i in sales:
    s1.add(i)
print(f'Distinct Sales:{s1}')

# Returning the second highest sales from the distinct sales
distinct_sales=[]
for i in s1:
    distinct_sales.append(i)
distinct_sales.sort(reverse=True)
print(f'Second Highest Distinct Sales: {distinct_sales[1]}')
```

```python
# Excercise 1.2: Calculate the sales matrics manually without using any built-in functions
sales = [1200, 1500, 800, 2200, 1750, 900, 3000]

# find the total sales
total_sales=float(0)
for i in sales:
    total_sales+=i
print(f'Total Sales: {total_sales}')

# find the average sales
# We can optimize the code by calculating the average sales in the same loop where we calculate the total sales.
# directly assigning the average sales to a variable instead of initializing it to 0 and then calculating it in a separate loop.
# average_sales=float(0)  

# we can find the count of sales
sales_length=0
for i in sales:
    sales_length+=1
print(f'Number of Sales: {sales_length}')

average_sales=total_sales/sales_length
print(f'Average Sales: {average_sales}')

# count the number of days when sales exceeded average sales
count=0
for i in sales:
    if i>average_sales:
        count+=1
# displaying the average sales and the number of days when sales exceeded average sales
print(f'Number of days with sales exceeding average: {count}')
```

```python
product_sales = {
    "Laptop": 45000,
    "Mobile": 28000,
    "Tablet": 15000,
    "Headphones": 5000,
    "Keyboard": 3500,
    "Monitor": 18000,
    "Mouse": 1500
}
```

```python
# Calculate total sales revenue manually
```

```python
total_sales_revenue=0
for product in product_sales:
    total_sales_revenue+=product_sales[product]
print(f'Total Sales Revenue :{total_sales_revenue}')
```

```python
# Calculate the average revenue per product
```

```python
# No of product
count=0
for product in product_sales:
    count+=1
print(f'No of Product :{count}')

# average revenue per product
average_per_product=total_sales_revenue/count
print(f'Average Revenue per product :{average_per_product:.2f}')
```

```python
# Highest-selling product
```

```python
highest_value_product=float('-inf')
product_name=''
for i in product_sales:
    if product_sales[i]>highest_value_product:
        highest_value_product=product_sales[i]
        product_name=i
print(f'Highest Value Product : {product_name}')
print(f'Product Price : {product_sales[product_name]}')
print(f'Highest Sales Revenue: {highest_value_product}')
    
```

```python
# Lowest Sales revenue
# find the first product in product sales
first_value=0
for i in product_sales:
    first_key=i
    first_value=product_sales[i]
    break
print(f'Let Suppose 1s Product is lowest revenue product:{first_key}, value : {first_value}')

lowest_revenue=product_sales[first_key]
product_name=first_key
for i in product_sales:
    if product_sales[i]<lowest_revenue:
       lowest_revenue=product_sales[i]
       product_name=i
print(f'The Lowest revenue product is {product_name}, {lowest_revenue}')
```

```python
# calculating sales
total_sales=0
for i in product_sales:
    total_sales+=product_sales[i]
print(f'Total Sales :{total_sales}')

# no of sales
sales_count=0
for i in product_sales:
    sales_count+=1
print(f"Sales Count :{sales_count}")

# average sales
average_sales=total_sales/sales_count
print(f'Average Sales :{average_sales:.2f}')

# above average sales count
above_avg_sales_dict={}
avg_sales_count=0
for i in product_sales:
    if product_sales[i]>average_sales:
        avg_sales_count+=1
        above_avg_sales_dict[i]=product_sales[i]
print(f'Above Avg Sales Dictionary :{above_avg_sales_dict}')
print(f'Avg Sales Count :{avg_sales_count}')
```

```python
# below average sales count
below_avg_sales_dict={}
for i in product_sales:
    if product_sales[i]<average_sales:
        below_avg_sales_dict[i]=product_sales[i]
print(f'Below Avg Sales Dictionary :{below_avg_sales_dict}')
```

```python
# Percentage contribution of above average sales in total sales revenue


# above average sales count
above_avg_sales_dict={}
avg_sales_count=0
for i in product_sales:
    if product_sales[i]>average_sales:
        avg_sales_count+=1
        above_avg_sales_dict[i]=product_sales[i]
print(f'Above Avg Sales Dictionary :{above_avg_sales_dict}')

# percentage contribution of each above average sales product
for i in above_avg_sales_dict:
    print(f'Sales contribution of {i} in {total_sales} is {(above_avg_sales_dict[i]/total_sales)*100:.2f}')
    
# percentage contribution of each product in total_sales revenue
for i in product_sales:
    print(f'Sales contribution of each product in total_sales revenue : {i} is {(product_sales[i]/total_sales)*100:.2f}')
```
