```python
# Project Mini E-Commerce Revenue Calculator
```

```python
product_name=input("Enter Product name: ")
product_price=float(input("Enter Product price: "))
product_quantity=int(input("Enter Product quantity: "))
product_discount=float(input("Enter Total Discount: "))

gross_revenue=product_price*product_quantity
discount_amount=(gross_revenue*product_discount)/100
final_revenue=gross_revenue-discount_amount

print("\n-- Store Invoice --")
print("Gross Revenue : ",gross_revenue)
print("Discount Amount : ",discount_amount)
print("Final Revenue : ",final_revenue)

print("\n-- Product Invoice --")
print("Product Name : ",product_name)
print("Product Quantity : ",product_quantity)
print("Total Amount : ",final_revenue)
```
